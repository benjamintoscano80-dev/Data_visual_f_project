```python
import geonamescache
import pandas as pd

gc = geonamescache.GeonamesCache()
cities = gc.get_cities()

# Normalizzazione in DataFrame
df_cities = pd.DataFrame.from_dict(cities, orient="index")[
    ["name", "countrycode", "latitude", "longitude", "population"]
]
countries = pd.DataFrame.from_dict(gc.get_countries(), orient="index")[
    ["iso", "name"]
].rename(columns={"name": "country_name"})

# Merge per avere il nome esteso del paese
df_mapping = df_cities.merge(
    countries, left_on="countrycode", right_on="iso", how="left"
)
```
