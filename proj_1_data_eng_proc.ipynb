```python
df_main = pd.concat([df_1, df_2, df_3, df_4], ignore_index=True)
df_main.duplicated().sum()
```

```python
df_main.drop(
    columns=["city_ascii", "iso2", "iso3", "ISO2_Code", "ISO3_Code", "id", "LocID", 'TimeMid', "Country Code", "Notes", 'SDMX_Code', "Index", "LocType", "LocTypeName", "ParentID"],
    inplace=True,
)
```

```python
gc = geonamescache.GeonamesCache()
cities = gc.get_cities()

# Normalizzazione in DataFrame
df_cities = pd.DataFrame.from_dict(cities, orient="index")[
    ["name", "countrycode", "latitude", "longitude", "population"]
]
countries = pd.DataFrame.from_dict(gc.get_countries(), orient="index")[
    ["iso", "name"]
].rename(columns={"name": "country"})

# Merge per avere il nome esteso del paese
df_mapping = df_cities.merge(
    countries, left_on="countrycode", right_on="iso", how="left"
)

df_mapping.rename(columns={"name": "city"}, inplace=True)

df_mapping = df_mapping[["city", "country"]]
```

```python
df_temp = df_main[["city", "country"]]

# Normalizza df_mapping
mapping_clean = df_mapping.drop_duplicates(subset=["city"]).copy()
mapping_clean["city_clean"] = mapping_clean["city"].astype(str).str.strip().str.lower()
city_country_map = mapping_clean.set_index("city_clean")["country"]

# Normalizza df_temp in fase di lookup
city_lookup = df_temp["city"].astype(str).str.strip().str.lower()

# Assegnazione diretta o tramite combine_first
df_temp["country"] = df_temp["country"].fillna(city_lookup.map(city_country_map))

# df_temp.loc[df_temp["city"] == "Aruba", "country"] = "Netherlands"

citta_senza_paese = df_temp.loc[df_temp["country"].isna(), "city"].unique()
paesi_senza_citta = df_temp.loc[df_temp["city"].isna(), "country"].unique()
```
